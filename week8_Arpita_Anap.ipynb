## WEEK 8 Build an Agentic AI pipeline

### Objective

Build a Short Quiz Agent Pipeline that can:

- Understand the user's query
- Decide which tool should be used
- Execute the appropriate tool
- Return the response in structured JSON format

### 1. Import Required Library 

```python
import json
```

### Step 1: Build Calculator Tool
The tool must solve mathematical expressions.

```python
def calculator(expression):
    """Evaluate a mathematical expression."""

    try:
        result = eval(expression)
        return str(result)

    except Exception:
        return "Error in calculation"
```

### Step 2: Build Keyword Extraction Tool

The tool extracts important keywords from a sentence.

```python
def extract_keywords(text):
    """Extract important words from the given text."""

    try:
        words = text.split()

        keywords = list(
            set(
                word.lower()
                for word in words
                if len(word) > 4
            )
        )

        return keywords[:5]

    except Exception:
        return []
```

### Step 3: General Response Handler

If a query does not require the calculator or keyword tool, the agent sends it to the general response handler.

```python
def general_response(query):
    """Handle normal conversational queries."""

    return "This is a general query. No specific tool is required."
```

#### Step 4: Agent Routing Logic

The agent checks the query in lowercase form.

Now Routing rules:

- "calculate" → Calculator
- "keywords" → Keyword Extractor
- Other queries → General Response
- Invalid input → Error

```python
# Agent function

def agent(query):

    # Check for invalid or empty input
    if not isinstance(query, str) or not query.strip():

        return {
            "type": "error",
            "result": "Query cannot be empty."
        }

    # Convert the query to lowercase
    query_lower = query.lower().strip()

    # Calculator route
    
    if "calculate" in query_lower:

        expression = query_lower.replace(
            "calculate",
            "",
            1
        ).strip()

        # No expression after "calculate"
        if not expression:

            return {
                "type": "error",
                "result": "No mathematical expression provided."
            }

        result = calculator(expression)

        # Calculator failed
        if result == "Error in calculation":

            return {
                "type": "error",
                "result": "Invalid mathematical expression."
            }

        return {
            "type": "calculation",
            "result": result
        }

    # Keyword Route

    elif "keywords" in query_lower:

        text = query

        # Remove common instruction words
        text = text.replace(
            "Extract keywords from",
            "",
            1
        )

        text = text.replace(
            "extract keywords from",
            "",
            1
        )

        text = text.strip()


        # No text provided
        if not text:

            return {
                "type": "error",
                "result": "No text provided for keyword extraction."
            }

        result = extract_keywords(text)

        return {
            "type": "keywords",
            "result": result
        }

    # General Route

    else:

        result = general_response(query)

        return {
            "type": "general",
            "result": result
        }
```

#### Step 5: Automated Agent Testing

Testing multiple examples for each routing condition.

```python
# AUTOMATED TESTING

test_queries = [

    "Calculate 125 * 8",

    "Calculate 500 / 25",

    "Extract keywords from Python is widely used in artificial intelligence and data science",

    "Find keywords from machine learning algorithms for prediction",

    "Tell me about cloud computing"

]


for query in test_queries:

    response = agent(query)

    print("\nQuery:", query)

    print("Response:")

    print(response)

    print("-" * 50)
```

#### Step 6: Validation Checks

Now Checking that every agent response follows the
required structure:

{
    "type": ...,
    "result": ...
}

```python
# Validation Check

print("\nVALIDATION RESULTS")
print("=" * 60)


for query in test_queries:

    response = agent(query)

    valid_response = (
        isinstance(response, dict)
        and "type" in response
        and "result" in response
    )

    print("\nQuery:", query)

    if valid_response:
        print("Status: PASS")
    else:
        print("Status: FAIL")

    print("Response:", response)
```

#### Step 7: Error Handling

To check whether the user gives an empty or incorrectly formatted command.

```python
# Error Testing

error_queries = [

    "",

    "Calculate",

    "Calculate 25 +",

    "Keywords"

]


for query in error_queries:

    response = agent(query)

    print("\nQuery:", repr(query))

    print("Response:", response)

    print("-" * 50)
```

#### Step 8: Interactive Agent

It allows user to communicate directly with the agent until "exit" is entered.

```python
# Interactive Mode

print("\n" + "=" * 60)

print("SINGLE-AGENT SMART ASSISTANT")

print("=" * 60)

print("You can ask calculations, extract keywords,")
print("or enter a general question.")

print("Type 'exit' to stop.\n")


while True:

    user_input = input("Enter query: ")


    # Stop the program
    if user_input.lower().strip() == "exit":

        print("\nAgent stopped.")

        break


    # Send the query to the agent
    response = agent(user_input)


    # Display structured response
    print("\nResponse:")

    print(response)

    print("-" * 50)
```
